```python
# Standardize user type values
df["user_type"] = (df["user_type"].astype("string").str.strip().str.title())
```

```python
#check user_type values

df["user_type"].value_counts()
```

```python
# List of columns containing text or category values
text_columns = [
    "event_name",
    "city",
    "state",
    "content_category",
    "plan",
    "payment_method",
    "failure_reason"
]
```

```python
# Remove unnecessary spaces from these columns
for column in text_columns:df[column] = df[column].astype("string").str.strip()
```

```python
# Check missing values after the cleaning
df.isnull().sum()
```

```python
# Replace missing values with "Unknown"
df["city"] = df["city"].fillna("Unknown")
df["payment_method"] = df["payment_method"].fillna("Unknown")
df["content_category"] = df["content_category"].fillna("Unknown")
```

```python
# Confirm the missing values have been handled
df.isnull().sum()
```

```python
# Check subscription plans
df["plan"].value_counts()
```

```python
# Check payment methods
df["payment_method"].value_counts()
```

```python
# Check why payments failed
df["failure_reason"].value_counts()
```

```python
# Arrange each user's events in chronological order
df = df.sort_values( by=["user_id", "event_timestamp"]).reset_index(drop=True)
```

```python
# Check the journey of one user
df[df["user_id"] == "U00001"]
```

```python
# Final check after cleaning

print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))
print("Unique users:", df["user_id"].nunique())
print("Duplicate rows:", df.duplicated().sum())

print("\nMissing values:")
print(df.isnull().sum())
```
